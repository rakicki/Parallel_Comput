```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Read CSV

df = pd.read_csv('sol_rank0.csv', header=None)
df.columns = ['x', 'y', 'z']

# Pivot to create grid if necessary
X_unique = np.sort(df['x'].unique())
Y_unique = np.sort(df['y'].unique())
X, Y = np.meshgrid(X_unique, Y_unique)
Z = df.pivot(index='y', columns='x', values='z').values  # make 2D array

# Surface plot
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, Z, cmap='viridis')
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
plt.show()

# Contour plot
plt.figure(figsize=(8,6))
plt.contourf(X, Y, Z, levels=50, cmap='viridis')
plt.colorbar()
plt.xlabel('X')
plt.ylabel('Y')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np


P = np.array([1, 2, 4, 8])
S_measured = np.array([1.0, 1.11, 1.09, 0.89])

S_ideal = P.copy()

plt.figure(figsize=(6, 4))
plt.plot(P, S_measured, 'o-', linewidth=2, markersize=8, label='Speedup mesuré')
plt.plot(P, S_ideal, '--', color='gray', linewidth=2, label='Speedup idéal')

plt.title("Courbe de Speedup — méthode de Jacobi (Nx=Ny=2000, 200 itérations)")
plt.xlabel("Nombre de processus (P)")
plt.ylabel("Speedup (S)")
plt.xticks(P)
plt.grid(True, which='both', linestyle='--', linewidth=0.5)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Read the .dat file with semicolon separator
data = pd.read_csv("Gauss_Sidel_err.dat", sep=";", header=None, names=["iteration", "error"])

# Extract columns
iterations = data["iteration"]
errors = data["error"]

# Plot error vs iteration (log scale on y-axis)
plt.semilogy(iterations, errors, marker='o', linestyle='-')
plt.xlabel("Itérations")
plt.ylabel("Erreur(échelle log)")
plt.title(" Convergence de la méthode de Gauss-Seidel ")
plt.grid(True, which="both")
plt.show()
```

```python


import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Read CSV

df = pd.read_csv('solution_globalci.csv', header=None)
df.columns = ['x', 'y', 'z']

# Pivot to create grid if necessary
X_unique = np.sort(df['x'].unique())
Y_unique = np.sort(df['y'].unique())
X, Y = np.meshgrid(X_unique, Y_unique)
Z = np.cos(np.pi*X)*np.cos(2*np.pi*Y)   # Example: f(x, y) = sin(sqrt(x^2 + y^2))

# Surface plot
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, Z, cmap='viridis')
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
plt.show()

# Contour plot
plt.figure(figsize=(8,6))
plt.contourf(X, Y, Z, levels=50, cmap='viridis')
plt.colorbar()
plt.xlabel('X')
plt.ylabel('Y')
plt.show()
```
