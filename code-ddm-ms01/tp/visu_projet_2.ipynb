```python
import numpy as np
import matplotlib.pyplot as plt
F = np.array([0.0025, 0.005, 0.01, 0.025, 0.05])
N_iteration = np.array([1089, 633, 347, 151, 92])

S_ideal = F.copy()

plt.figure(figsize=(6, 4))
plt.plot(F, N_iteration, 'o-', linewidth=2, markersize=8)

plt.title("Nombre d'itérations nécessaire pour atteindre une erreur relative en norme H¹ de 1e-6")
plt.xlabel("Taille du maillage (h)")
plt.ylabel("Nombre d'itérations")
plt.xticks(F)
plt.grid(True, which='both', linestyle='--', linewidth=0.5)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

x = []
y = []

with open("cpu_time.dat") as f:
    for line in f:
        i, t = map(float, line.split())
        x.append(i)
        y.append(t)

plt.plot(x, y)
plt.xlabel("Number of points")
plt.ylabel("CPU time (s)")
plt.title("CPU Time vs Number of Inserted Points")
plt.grid()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Read the .dat file with semicolon separator
data025 = pd.read_csv("TP3_exo3_h0.025.dat", sep=";", header=None, names=["iteration", "error"])
data005 = pd.read_csv("TP3_exo3_h0.005.dat", sep=";", header=None, names=["iteration", "error"])
data01 = pd.read_csv("TP3_exo3_h0.01.dat", sep=";", header=None, names=["iteration", "error"])
data0025 = pd.read_csv("TP3_exo3_h0.0025.dat", sep=";", header=None, names=["iteration", "error"])
# Extract columns
iterations025 = data025["iteration"]
errors025 = data025["error"]
iterations005 = data005["iteration"]
errors005 = data005["error"]
iterations01 = data01["iteration"]
errors01 = data01["error"]
iterations0025 = data0025["iteration"]
errors0025 = data0025["error"]

# Plot error vs iteration (log scale on y-axis)
plt.semilogy(iterations025, errors025, marker='o', linestyle='-', label='h=0.025')
plt.semilogy(iterations005, errors005, marker='o', linestyle='-', label='h=0.005')
plt.semilogy(iterations01, errors01, marker='o', linestyle='-', label='h=0.01')
plt.semilogy(iterations0025, errors0025, marker='o', linestyle='-', label='h=0.0025')
plt.xlabel("Iteration")
plt.ylabel("Erreur en norme H^1 (log scale) ")
plt.title("Erreur relative en norme euclidienne solver gradient conjugué  ASM")
plt.grid(True, which="both")
plt.legend()
plt.show()
```
